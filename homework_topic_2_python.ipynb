## Задание 1. Високосный или обычный год

Год считается високосным, если он делится на 400 **или** делится на 4, но при этом не делится на 100.

```python
year = 2024

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print('Високосный год')
else:
    print('Обычный год')
```

### Проверка на 2023 год

Аналогично меняем значение переменной на `2023`.

```python
year = 2023

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print('Високосный год')
else:
    print('Обычный год')
```

## Задание 2 (дополнительное). Счастливый билет

Сравниваем сумму первых трёх цифр с суммой последних трёх цифр шестизначного номера.

```python
number = 123456

digits = str(number)
first_sum = sum(int(digit) for digit in digits[:3])
last_sum = sum(int(digit) for digit in digits[3:])

if first_sum == last_sum:
    print('Счастливый билет')
else:
    print('Несчастливый билет')
```

### Проверка на 123321

```python
number = 123321

digits = str(number)
first_sum = sum(int(digit) for digit in digits[:3])
last_sum = sum(int(digit) for digit in digits[3:])

if first_sum == last_sum:
    print('Счастливый билет')
else:
    print('Несчастливый билет')
```

**Результаты самопроверки:** `2024` → «Високосный год»; `2023` → «Обычный год»; `123456` → «Несчастливый билет»; `123321` → «Счастливый билет».
