**Задача:** используя `purchase_log.txt`, сформировать файл `funnel.csv` со столбцами `user_id, source, category` только для визитов с покупками. Исходный `visit_log.csv` обрабатывается построчно.

В GitHub разместите в одной папке три файла: этот ноутбук, архив `Материалы_к_ДЗ_5(2).zip` и получившийся `funnel.csv`. Запускайте ноутбук в Jupyter Notebook (GitHub только показывает файлы).

```python
import csv
import io
import json
import zipfile

# ZIP-архив лежит рядом с этим ноутбуком
archive_name = 'Материалы_к_ДЗ_5(2).zip'

with zipfile.ZipFile(archive_name) as archive:
    # Словарь покупок: user_id -> категория
    purchases = {}
    with archive.open('purchase_log.txt') as file:
        for line in io.TextIOWrapper(file, encoding='utf-8'):
            purchase = json.loads(line)
            if purchase['user_id'] != 'user_id':
                purchases[purchase['user_id']] = purchase['category']

    # visit_log.csv читаем только построчно
    with archive.open('visit_log.csv') as source, open(
        'funnel.csv', 'w', encoding='utf-8', newline=''
    ) as target:
        reader = csv.DictReader(io.TextIOWrapper(source, encoding='utf-8'))
        writer = csv.writer(target)
        writer.writerow(['user_id', 'source', 'category'])

        for visit in reader:
            category = purchases.get(visit['user_id'])
            if category is not None:
                writer.writerow([visit['user_id'], visit['source'], category])

print('Файл funnel.csv создан')
```
