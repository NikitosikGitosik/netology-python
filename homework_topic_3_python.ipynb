## Задание 1. Средние буквы слова

Если количество букв нечётное, выводим одну среднюю букву. Если чётное — две средние буквы.

```python
word = 'test'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

**Проверка второго примера** — `word = 'testing'`.

```python
word = 'testing'

middle = len(word) // 2

if len(word) % 2 == 0:
    print(word[middle - 1:middle + 1])
else:
    print(word[middle])
```

## Задание 2 (необязательное). Идеальные пары

Сначала проверяем, одинаковое ли количество юношей и девушек. Если да — сортируем оба списка и объединяем имена по порядку.

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```

**Проверка второго примера** — юношей больше, чем девушек.

```python
boys = ['Peter', 'Alex', 'John', 'Arthur', 'Richard', 'Michael']
girls = ['Kate', 'Liza', 'Kira', 'Emma', 'Trisha']

if len(boys) != len(girls):
    print('Внимание, кто-то может остаться без пары!')
else:
    print('Идеальные пары:')
    for boy, girl in zip(sorted(boys), sorted(girls)):
        print(boy, 'и', girl)
```
