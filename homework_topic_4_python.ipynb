## Задача
Автоматизировать работу секретаря с документами:

- `p` — найти владельца документа по номеру (обязательная часть);
- `s` — определить полку по номеру (дополнительная часть);
- `q` — завершить программу.

Программа использует функции для отдельных задач и цикл `while`, завершающийся только по команде `q`.


## 1. Исходные данные

```python
documents = [
    {'type': 'passport', 'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice', 'number': '11-2', 'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006', 'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}
```

## 2. Функции для поиска документа и полки

Каждая функция отвечает за одну задачу. Поиск владельца переиспользует функцию поиска документа.

```python
def find_document(document_number):
    """Найти документ по номеру или вернуть None."""
    for document in documents:
        if document['number'] == document_number:
            return document
    return None


def get_document_owner(document_number):
    """Вернуть сообщение с именем владельца документа."""
    document = find_document(document_number)
    if document is None:
        return 'Владелец документа: владелец не найден'
    return f"Владелец документа: {document['name']}"


def get_document_shelf(document_number):
    """Вернуть сообщение с номером полки документа."""
    for shelf_number, document_numbers in directories.items():
        if document_number in document_numbers:
            return f'Документ хранится на полке: {shelf_number}'
    return 'Документ не найден'
```

## 3. Проверка по чек-листу

Ниже представлены все четыре проверочных случая из задания. Если проверки пройдены, выводятся нужные результаты.

```python
assert get_document_owner('10006') == 'Владелец документа: Аристарх Павлов'
assert get_document_owner('1001') == 'Владелец документа: владелец не найден'
assert get_document_shelf('10006') == 'Документ хранится на полке: 2'
assert get_document_shelf('1001') == 'Документ не найден'

print(get_document_owner('10006'))
print(get_document_owner('1001'))
print(get_document_shelf('10006'))
print(get_document_shelf('1001'))
```

## 4. Основная программа

Функция `main()` обрабатывает команды в цикле `while` до ввода `q`. Для работы с программой запустите **последнюю ячейку**, вводите команду `p`, `s` или `q` и, если нужно, номер документа.

```python
def main():
    while True:
        command = input('Введите команду: ').strip().lower()

        if command == 'q':
            print('Программа завершена.')
            break
        elif command == 'p':
            document_number = input('Введите номер документа: ').strip()
            print(get_document_owner(document_number))
        elif command == 's':
            document_number = input('Введите номер документа: ').strip()
            print(get_document_shelf(document_number))
        else:
            print('Неизвестная команда. Используйте p, s или q.')


main()
```
